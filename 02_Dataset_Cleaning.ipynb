```python
import os
import hashlib
from PIL import Image
from tqdm import tqdm
import pandas as pd
```

```python
DATASET_PATH = r"C:\Users\sujan\Desktop\final project2\DATASET\RiceDiseaseDataset"
```

```python
#Find Corrupted Images
corrupted_images = []

for class_name in os.listdir(DATASET_PATH):

    class_path = os.path.join(DATASET_PATH, class_name)

    if not os.path.isdir(class_path):
        continue

    for image_name in tqdm(os.listdir(class_path), desc=class_name):

        image_path = os.path.join(class_path, image_name)

        try:
            with Image.open(image_path) as img:
                img.verify()

        except Exception:
            corrupted_images.append(image_path)

print(f"\nCorrupted Images Found : {len(corrupted_images)}")
```

```python
#Remove Corrupted Images
for image in corrupted_images:
    os.remove(image)

print("Corrupted images removed successfully.")
```

```python
#Find Duplicate Images
hash_dict = {}
duplicate_images = []

for class_name in os.listdir(DATASET_PATH):

    class_path = os.path.join(DATASET_PATH, class_name)

    if not os.path.isdir(class_path):
        continue

    for image_name in tqdm(os.listdir(class_path), desc=class_name):

        image_path = os.path.join(class_path, image_name)

        with open(image_path, "rb") as f:
            file_hash = hashlib.md5(f.read()).hexdigest()

        if file_hash in hash_dict:
            duplicate_images.append(image_path)
        else:
            hash_dict[file_hash] = image_path

print(f"\nDuplicate Images Found : {len(duplicate_images)}")
```

```python
#Remove Duplicate Images
for image in duplicate_images:
    os.remove(image)

print("Duplicate images removed successfully.")
```

```python
# Cleaning Summary
print("=" * 60)
print("DATASET CLEANING SUMMARY")
print("=" * 60)

print(f"Corrupted Images Removed : {len(corrupted_images)}")
print(f"Duplicate Images Removed : {len(duplicate_images)}")

print("\nDataset is now clean and ready for splitting.")
```

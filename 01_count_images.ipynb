```python
import os
```

```python
DATASET_PATH = r"C:\Users\sujan\Desktop\final project2\DATASET\RiceDiseaseDataset"

# Supported image formats
IMAGE_EXTENSIONS = (".jpg", ".jpeg", ".png", ".bmp", ".tif", ".tiff", ".webp")

print("=" * 70)
print("          RICE DISEASE DATASET ANALYSIS")
print("=" * 70)

total_images = 0
total_non_images = 0

print(f"\nDataset Location:\n{DATASET_PATH}\n")
```

```python
# Get class folders
classes = sorted(
    [folder for folder in os.listdir(DATASET_PATH)
     if os.path.isdir(os.path.join(DATASET_PATH, folder))]
)

print(f"Number of Classes : {len(classes)}\n")

print("-" * 70)
print(f"{'Class Name':30} {'Images':>10} {'Non-Images':>15}")
print("-" * 70)

for class_name in classes:

    class_path = os.path.join(DATASET_PATH, class_name)

    image_count = 0
    non_image_count = 0

    for file in os.listdir(class_path):

        file_path = os.path.join(class_path, file)

        if os.path.isfile(file_path):

            if file.lower().endswith(IMAGE_EXTENSIONS):
                image_count += 1
            else:
                non_image_count += 1

    total_images += image_count
    total_non_images += non_image_count

    print(f"{class_name:30} {image_count:10} {non_image_count:15}")

print("-" * 70)
print(f"{'TOTAL':30} {total_images:10} {total_non_images:15}")
print("-" * 70)

print("\nAnalysis Completed Successfully.")
```
